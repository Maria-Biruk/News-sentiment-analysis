```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
news = pd.read_csv("../data/raw/newsData/raw_analyst_ratings.csv")

news.head()
```

```python
news.info()

news.isnull().sum()
```

```python
news['headline_length'] = news['headline'].astype(str).apply(len)

news[['headline', 'headline_length']].head()
```

```python
news['date'] = pd.to_datetime(news['date'], errors='coerce')

news['year'] = news['date'].dt.year
news['month'] = news['date'].dt.month
```

```python
top_publishers = news['publisher'].value_counts().head(10)

top_publishers
```

```python
plt.figure(figsize=(10,5))
top_publishers.plot(kind='bar')
plt.title("Top News Publishers")
plt.show()
```

```python
top_stocks = news['stock'].value_counts().head(10)

top_stocks
```
