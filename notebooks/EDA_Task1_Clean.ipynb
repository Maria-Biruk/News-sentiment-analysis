```python
news = pd.read_csv("../data/raw/newsData/raw_analyst_ratings.csv")
stock = pd.read_csv("../data/raw/yfinance_data/data/AAPL.csv")
```

```python
news.isnull().sum()
```

```python
# remove duplicates
news = news.drop_duplicates()

# remove rows where headline is missing (very important for NLP)
news = news.dropna(subset=['headline'])

# check again
news.isnull().sum()
```

```python
news.info()
news.head()
```

```python
news['headline_length'] = news['headline'].astype(str).apply(len)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10,5))
sns.histplot(news['headline_length'], bins=50)
plt.title("Distribution of News Headline Length")
plt.show()
```

Most headlines are short in length, indicating that financial news is typically concise and structured for quick consumption.

```python
news['publisher'].value_counts().head(10)
```

```python
top_publishers = news['publisher'].value_counts().head(10)

plt.figure(figsize=(10,5))
top_publishers.plot(kind='bar')
plt.title("Top 10 News Publishers")
plt.show()
```

A small number of publishers dominate financial news production, indicating centralized news distribution.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
top_publishers.plot(kind='bar')
plt.title("Top 10 News Publishers")
plt.show()
```

A small number of publishers dominate financial news reporting. This suggests that financial news distribution is centralized, with a few major sources producing most headlines.

```python
news['date'] = pd.to_datetime(news['date'], errors='coerce')
```

```python
daily_news = news.groupby(news['date'].dt.date).size()
daily_news.head()
```

```python
plt.figure(figsize=(12,5))
plt.plot(daily_news.index, daily_news.values)
plt.title("Daily News Volume Over Time")
plt.show()
```

News activity fluctuates over time, indicating that financial reporting is event-driven rather than constant.

Large-cap stocks dominate news coverage, showing that media attention is concentrated on high-impact companies.
